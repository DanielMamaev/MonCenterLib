```python
from moncenterlib.gnss.quality_check import Anubis
anubis = Anubis()
```

### Basic examples

#### Scan directories

```python
dir_obs = "./ExampleData4MonCenterLib/quality_check/Anubis/obs"
dir_nav = "./ExampleData4MonCenterLib/quality_check/Anubis/nav"

#without recursion find
result = anubis.scan_dirs(dir_obs, dir_nav, False)
print(result)

#with recursion find
result = anubis.scan_dirs(dir_obs, dir_nav, True)
print(result)
```

#### Start processing

##### input from method scan_dirs

```python
files, no_match = anubis.scan_dirs(dir_obs, dir_nav, False)
result = anubis.start(files)
result
```

##### input dirs

```python
input_dirs = (dir_obs, dir_nav)

#without recursion find
result = anubis.start(input_dirs, False)
print(result)

#with recursion find
result = anubis.start(input_dirs, True)
print(result)
```

##### input one file

```python
input_files = ("./ExampleData4MonCenterLib/quality_check/Anubis/obs/novm0060.22o",
               "./ExampleData4MonCenterLib/quality_check/Anubis/nav/BRDC00IGS_R_20220060000_01D_MN.rnx")
result = anubis.start(input_files)
result
```

##### Saving xtr files in another directory

```python
!mkdir output_xtr

input_dirs = (dir_obs, dir_nav)
result = anubis.start(input_dirs, True, "./output_xtr")
result
```

#### Disable logger

```python
anubis = Anubis(False)
input_dirs = (dir_obs, dir_nav)
result = anubis.start(input_dirs, False, "/content/output_xtr")
result
```
