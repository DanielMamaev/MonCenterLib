```python
from moncenterlib.gnss.cddis_client import CDDISClient
cddis_cli = CDDISClient()
```

### Basic examples

#### Download daily multignss boardcast ephemeris

```python
# create directory - multi_gnss_brd_eph
!mkdir multi_gnss_brd_eph

query = {
    "start": "2020-12-30",
    "end": "2021-01-02"
    }
# or list
# query = ["2020-12-30", "2020-05-05"]

result = cddis_cli.get_daily_multi_gnss_brd_eph("./multi_gnss_brd_eph", query)
result
```

#### Download observation daily 30 second data

```python
# create directory - daily_30s_data
!mkdir daily_30s_data

query = {
    "dates": {"start": "2020-12-30", "end": "2021-01-02"},
    "station": "NOVM",
    "type": "O",
    "rinex_v": "auto"
    }

# or list of dates
query = {
    "dates": ["2020-12-30", "2020-05-05"],
    "station": "NOVM",
    "type": "O",
    "rinex_v": "auto"
    }

result = cddis_cli.get_daily_30s_data("./daily_30s_data", query)
result
```

#### Download ephemeris daily 30 second data

```python
# create directory - daily_30s_data_eph
!mkdir daily_30s_data_eph

# rinex version 3
query = {
    "dates": {"start": "2020-12-30", "end": "2021-01-02"},
    "station": "GODN",
    "type": "N",
    "rinex_v": "3"
    }

result = cddis_cli.get_daily_30s_data("./daily_30s_data_eph", query)
print(result)

# rinex version 2
query = {
    "dates": {"start": "2020-12-30", "end": "2021-01-02"},
    "station": "GODN",
    "type": "N",
    "rinex_v": "2"
    }

result = cddis_cli.get_daily_30s_data("./daily_30s_data_eph", query)
print(result)
```

#### Download precise orbits

```python
# create directory - precise_orbits
!mkdir precise_orbits

query = {
    "start": "2020-12-30",
    "end": "2021-01-02"
    }

# or list
# query = ["2020-12-30", "2020-05-05"]

result = cddis_cli.get_precise_orbits("./precise_orbits", query)
print(result)
result = cddis_cli.get_precise_orbits("./precise_orbits", query, glonass=True)
print(result)
```

### Other features

```python
# download only archive

# create directory - archive
!mkdir archive

query = {
    "start": "2020-12-30",
    "end": "2020-12-30"
}

result = cddis_cli.get_clock_5m("./archive", query, unpack=False)
result = cddis_cli.get_clock_30s("./archive", query, unpack=False)
result = cddis_cli.get_precise_orbits("./archive", query, unpack=False)
result = cddis_cli.get_earth_orientation("./archive", query, unpack=False)
result = cddis_cli.get_daily_multi_gnss_brd_eph("./archive", query, unpack=False)

query = {
    "dates": {"start": "2022-12-30", "end": "2022-12-30"},
    "station": "GODN",
    "type": "O",
    "rinex_v": "auto"
    }

result = cddis_cli.get_daily_30s_data("./archive", query, unpack=False)
print(result)
```
