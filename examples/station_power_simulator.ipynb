```python
start_date = "2025-01-01"
end_date = "2026-01-01"

latitude = 54.9874                                
longitude = 82.8646

timezone = "Asia/Novosibirsk"

config_solar_panel_135 = {
    "latitude": latitude,                               # Your coordinates
    "longitude": longitude,                             # Your coordinates

    "start_date": start_date,                           # Date range YYYY-MM-DD
    "end_date": end_date,                               # Date range YYYY-MM-DD
    
    "tilt_deg": 90,                                     # Tilt angle of the solar panel from horizontal (degrees). Range 0-90
    "panel_azimuth_deg": 135,                           # Panel azimuth: 180° = south, 90° = east, 270° = west, 0° = north.

    "area_m2": 0.627,                                   # Total panel area (m²).
    "panel_efficiency": 0.1923,                         # Module conversion efficiency
    "dc_efficiency": 1,                                 # Electrical efficiency of DC system (controller, wiring).

    "timezone": timezone,                               # Time zone of the station location (IANA format).
    # "timezone": ""                                    # Leave empty ("") to detect timezone automatically from coordinates.
}

config_solar_panel_225 = {
    "latitude": latitude,                               # Your coordinates
    "longitude": longitude,                             # Your coordinates

    "start_date": start_date,                           # Date range YYYY-MM-DD
    "end_date": end_date,                               # Date range YYYY-MM-DD
    
    "tilt_deg": 90,                                     # Tilt angle of the solar panel from horizontal (degrees). Range 0-90
    "panel_azimuth_deg": 225,                           # Panel azimuth: 180° = south, 90° = east, 270° = west, 0° = north.

    "area_m2": 0.627,                                   # Total panel area (m²).
    "panel_efficiency": 0.1923,                         # Module conversion efficiency
    "dc_efficiency": 1,                                 # Electrical efficiency of DC system (controller, wiring).

    "timezone": timezone,                               # Time zone of the station location (IANA format).
    # "timezone": ""                                    # Leave empty ("") to detect timezone automatically from coordinates.
}

config_wind_turbine = {
    "latitude": latitude,                               # Your coordinates
    "longitude": longitude,                             # Your coordinates

    "start_date": start_date,                           # Date range YYYY-MM-DD
    "end_date": end_date,                               # Date range YYYY-MM-DD

    "diameter": 0.66,                                   # Rotor diameter (m)
    "blade_length": 1.5,                                # Blade length (m). If you have a Vertical Axis (VAWT), set blade_length > 0. 
    # "blade_length": 0,                                # If you have a Horizontal Axis (HAWT), set blade_length = 0
    "rated_power": 450,                                 # Nominal power (W)
    "rated_speed": 11,                                  # Nominal wind speed (m/s)

    "cut_in": 1.3,                                      # Starting wind speed (m/s)
    "cut_out": 45,                                      # Maximum wind speed (m/s)
    
    "cp": 0.25,                                         # Aerodynamic power coefficient of rotor (Betz limit ≈ 0.59). Range 0.25 – 0.45
    "turbine_efficiency": 0.8,                          # Mechanical efficiency: gearbox, bearings, generator. Range 0.7 – 0.9
    "dc_efficiency": 1,                                 # Electrical efficiency of DC conversion (controller / wiring losses). Range 0.7 – 0.95

    "timezone": timezone,                               # Time zone of the station location (IANA format).
    # "timezone": ""                                    # Leave empty ("") to detect timezone automatically from coordinates.
}

solar_power_135 = SolarPanelPower(config_solar_panel_135)       # Create object
solar_power_135.calculate()                                     # Start calculate

solar_power_225 = SolarPanelPower(config_solar_panel_225)       # Create object
solar_power_225.calculate()                                     # Start calculate

wind_turbine = WindTurbinePower(config_wind_turbine)            # Create object
wind_turbine.calculate()                                        # Start calculate
```

```python
power_balance = PowerBalanceAnalyzer()                                                              # Create object
df_station = power_balance.combine_power([solar_power_135, solar_power_225, wind_turbine])          # combine powers
df_station
```

```python
power_load = 1.5 # Watt 
df, metrics = power_balance.calc_power_balance(df_station, power_load, path_protocol="power_balance_with_load.txt")   # power balance with load
```

```python
df_sim, metrics = power_balance.calc_power_balance_with_battery(                # power balance with load + battery
    input_df=df_station,            # DataFrame returned by combine_power()
    load_power_w=10.5,               # Constant load power (W)
    battery_capacity_Wh=35,         # Battery capacity (Wh)
    initial_soc_Wh=35,              # Initial battery state of charge (Wh)
    charge_efficiency=0.85,         # Battery charging efficiency
    discharge_efficiency=0.85,      # Battery discharging efficiency
    min_soc_Wh=5,                    # minimum battery state of charge (Wh)
    path_protocol = "power_balance_with_load_battery.txt"
)  


df_sim              # Simulation time-series DataFrame
metrics             # Summary metrics of the simulation
metrics["events"]   # List of timestamps when the system turned off/on
```
