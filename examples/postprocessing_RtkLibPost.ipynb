## Start postprocessing

### Single

```python
# create directory for .pos files
!mkdir single_pos

single_config = rtklib_post.get_default_config()
single_config["pos1-posmode"] = "0"
single_config["pos1-frequency"] = "2"
single_config["pos1-soltype"] = "2"
single_config["pos1-elmask"] = "15"
single_config["pos1-tidecorr"] = "0"
single_config["pos1-ionoopt"] = "1"
single_config["pos1-tropopt"] = "1"
single_config["pos1-sateph"] = "0"
single_config["pos1-navsys"] = "1"

single_config["out-solformat"] = "0"
single_config["out-timesys"] = "0"
single_config["out-timeform"] = "1"
single_config["out-degform"] = "0"

one_file = {
    "rover": "./ExampleData4MonCenterLib/postprocessing/rover/2022-01-04/nsk10040.22o",
    "nav": "./ExampleData4MonCenterLib/postprocessing/nav/2022004/BRDC00IGS_R_20220040000_01D_MN.rnx"
}
rtklib_post.start_single_processing(one_file, "./single_pos", single_config)
rtklib_post.output_files
```

```python
files = {
    "rover": "./ExampleData4MonCenterLib/postprocessing/rover",
    "nav": "./ExampleData4MonCenterLib/postprocessing/nav"
}

match_files, no_match = rtklib_post.match_files(files)

# see output files in dir single_pos
rtklib_post.start_multi_processing(match_files, "./single_pos", single_config, workers=4)
while True:
    time.sleep(1)
    status = rtklib_post.get_last_status()
    print(status)
    states = [s['isStop'] for s in status.values()]
    if all(states):
        break
print(rtklib_post.output_files)
```

### PPP static

```python
# create directory for .pos files
!mkdir ppp_pos



# set config
ppp_static_config = rtklib_post.get_default_config()
ppp_static_config["pos1-posmode"] = "7"
ppp_static_config["pos1-frequency"] = "2"
ppp_static_config["pos1-soltype"] = "2"
ppp_static_config["pos1-elmask"] = "15"
ppp_static_config["pos1-tidecorr"] = "2"
ppp_static_config["pos1-ionoopt"] = "3"
ppp_static_config["pos1-tropopt"] = "3"
ppp_static_config["pos1-sateph"] = "1"
ppp_static_config["pos1-posopt1"] = "1"
ppp_static_config["pos1-posopt2"] = "1"
ppp_static_config["pos1-posopt3"] = "1"
ppp_static_config["pos1-posopt4"] = "1"
ppp_static_config["pos1-posopt5"] = "1"
ppp_static_config["pos1-posopt6"] = "1"
ppp_static_config["pos1-navsys"] = "1"

ppp_static_config["pos2-armode"] = "0"

ppp_static_config["out-solformat"] = "0"
ppp_static_config["out-timesys"] = "0"
ppp_static_config["out-timeform"] = "1"
ppp_static_config["out-degform"] = "0"

ppp_static_config["ant1-anttype"] = "*"
ppp_static_config["ant1-antdele"] = "0"
ppp_static_config["ant1-antdeln"] = "0"
ppp_static_config["ant1-antdelu"] = "0"

ppp_static_config["file-satantfile"] = "./ExampleData4MonCenterLib/postprocessing/igs20.atx"
ppp_static_config["file-rcvantfile"] = "./ExampleData4MonCenterLib/postprocessing/igs20.atx"


# If you specified a file for some reason .erp in the configuration, specify erp_from_config=True
files = {
    "rover": "./ExampleData4MonCenterLib/postprocessing/rover/2022-01-04/nsk10040.22o",
    "nav": "./ExampleData4MonCenterLib/postprocessing/nav/2022004/BRDC00IGS_R_20220040000_01D_MN.rnx",
    "sp3": "./ExampleData4MonCenterLib/postprocessing/sp3/21912/igs21912.sp3",
    "clk": "./ExampleData4MonCenterLib/postprocessing/clk/21912/igs21912.clk_30s",
    "erp": "./ExampleData4MonCenterLib/postprocessing/erp/igs21917.erp"
}

print(rtklib_post.start_single_processing(files, "./ppp_pos", ppp_static_config))

```

```python
# scan dirs
dirs = {
    "rover": "./ExampleData4MonCenterLib/postprocessing/rover",
    "nav": "./ExampleData4MonCenterLib/postprocessing/nav",
    "sp3": "./ExampleData4MonCenterLib/postprocessing/sp3",
    "clk": "./ExampleData4MonCenterLib/postprocessing/clk",
    "erp": "./ExampleData4MonCenterLib/postprocessing/erp"
}
match_files, no_match = rtklib_post.match_files(dirs)

# An important point.
# If you have a folder with several files .erp and you pass a list of files .erp to the start() method, specify erp_from_config=False.
rtklib_post.start_multi_processing(match_files, "./ppp_pos", ppp_static_config, workers=4)
while True:
    time.sleep(1)
    status = rtklib_post.get_last_status()
    print(status)
    states = [s['isStop'] for s in status.values()]
    if all(states):
        break
print(rtklib_post.output_files)
```
