```python
from moncenterlib.gnss.rgs_client import RGSClient
```

### Basic examples

```python
rgs_cli = RGSClient(ssl=False) # use ssl=False if you see problem with SSL
```

#### Get information about station

```python
rgs_cli.get_station_info("NSK1")
```

#### Get informations about all stations

```python
rgs_cli.get_all_stations_info()
```

#### Download files

##### 1 file

```python
# create directory - one_file
!mkdir one_file

# date
filter_param = {
    "working_center": "NSK1",
    "date": "2020-01-01",
    "type": "O",
}
result = rgs_cli.download_files("./one_file", filter_param)
print(result)

# year and day
filter_param = {
    "working_center": "NSK1",
    "year": "2020",
    "day": "2",
    "type": "O",
}
result = rgs_cli.download_files("./one_file", filter_param)
print(result)

# gps_week and gps_week_day
filter_param = {
    "working_center": "NSK1",
    "gps_week": "2086",
    "gps_week_day": "5",
    "type": "O",
}
result = rgs_cli.download_files("./one_file", filter_param)
print(result)

# other type file
filter_param = {
    "working_center": "NSK1",
    "date": "2020-01-01",
    "type": "N",
}
result = rgs_cli.download_files("./one_file", filter_param)
print(result)
```

##### 2 and more files

```python
# create directory - more_files_date
!mkdir more_files_date

# date
filter_param = {
    "working_center": "NSK1",

    "date[0]": "2020-01-01",
    "date[1]": "2020-01-02",
    "date[2]": "2020-01-03",

    "type[0]": "O",
    "type[1]": "N",
}
result = rgs_cli.download_files("./more_files_date", filter_param)
print(result)

filter_param = {
    "working_center": "NSK1",

    "date[from]": "2020-01-04",
    "date[to]": "2020-01-06",

    "type": "O"
}
result = rgs_cli.download_files("./more_files_date", filter_param)
print(result)
```

```python
# create directory - year-day
!mkdir more_files_year-day

# year and day
filter_param = {
    "working_center": "NSK1",

    "year[0]": "2020",
    "year[1]": "2021",

    "day[0]": "10",
    "day[1]": "11",

    "type": "O",
}
result = rgs_cli.download_files("./more_files_year-day", filter_param)
print(result)

filter_param = {
    "working_center": "NSK1",

    "year[from]": "2018",
    "year[to]": "2020",

    "day[from]": "22",
    "day[to]": "23",

    "type": "O",
}
result = rgs_cli.download_files("./more_files_year-day", filter_param)
print(result)

filter_param = {
    "working_center": "NSK1",

    "year[from]": "2018",
    "year[to]": "2020",

    "day[0]": "32",
    "day[1]": "33",

    "type": "O",
}
result = rgs_cli.download_files("./more_files_year-day", filter_param)
print(result)
```

```python
# create directory - gps_week
!mkdir more_files_gps_week

# gps_week and gps_week_day
filter_param = {
    "working_center": "NSK1",

    "gps_week[0]": "2086",
    "gps_week[1]": "2088",

    "gps_week_day[0]": "5",
    "gps_week_day[1]": "1",

    "type": "O",
}
result = rgs_cli.download_files("./more_files_gps_week", filter_param)
print(result)

filter_param = {
    "working_center": "NSK1",

    "gps_week[from]": "2086",
    "gps_week[to]": "2087",

    "gps_week_day[from]": "1",
    "gps_week_day[to]": "2",

    "type": "O",
}
result = rgs_cli.download_files("./more_files_gps_week", filter_param)
print(result)
```

### Other feature

```python
# download only archive

# create directory - archive
!mkdir archive

filter_param = {
    "working_center": "NSK1",
    "date": "2020-01-01",
    "type": "O"
}
result = rgs_cli.download_files("./archive", filter_param, unpack=False)
print(result)
```
