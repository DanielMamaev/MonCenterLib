```python
from moncenterlib.gnss.tools4rnx import RtkLibConvbin
import time
t4r = RtkLibConvbin()
```

### Basic examples

#### Get default config

```python
config = t4r.get_default_config()
config
```

#### Scan directories

```python
#without recursion find
result = t4r.scan_dir("./ExampleData4MonCenterLib/tools4rnx/RtkLibConvbin", False)
print(result)

#without recursion find
result = t4r.scan_dir("./ExampleData4MonCenterLib/tools4rnx/RtkLibConvbin", True)
print(result)
```

#### Start processing

##### input from method scan_dir

```python
config = t4r.get_default_config()
config["format"] = "ubx"
files = t4r.scan_dir("./ExampleData4MonCenterLib/tools4rnx/RtkLibConvbin", False)
result = t4r.start(files, "./output_rnx", config)
while True:
    time.sleep(1)
    status = t4r.get_last_status()
    print(status)
    states = [s['isStop'] for s in status.values()]
    if all(states):
        break
print(t4r.output_files)
```

Multiprocessing

```python
t4r = RtkLibConvbin(workers=4)
config = t4r.get_default_config()
config["format"] = "ubx"
files = t4r.scan_dir("./ExampleData4MonCenterLib/tools4rnx/RtkLibConvbin", True)
result = t4r.start(files, "./output_rnx", config)
while True:
    time.sleep(1)
    status = t4r.get_last_status()
    print(status)
    states = [s['isStop'] for s in status.values()]
    if all(states):
        break
print(t4r.output_files)
```

##### input dir

```python
config = t4r.get_default_config()
config["format"] = "ubx"
result = t4r.start("./ExampleData4MonCenterLib/tools4rnx/RtkLibConvbin", "./output_rnx", config, True)
while True:
    time.sleep(1)
    status = t4r.get_last_status()
    print(status)
    states = [s['isStop'] for s in status.values()]
    if all(states):
        break
print(t4r.output_files)
```
