```python
from moncenterlib.stream2file import Stream2File

import time
# for pretty print
from pprint import pprint
```

### Work scenarios

#### Disable logger

```python
# logger off
s2f = Stream2File(False)

# logger on
s2f = Stream2File()
```

#### Add and remove connections

```python
param = {
    "type": "tcpcli",
    "addr": "igs.geomonlab.ru",
    "port": "6002"
    }


pprint("Empty connections")
pprint(s2f.get_connection_names())


pprint("ADD TEST")
s2f.add_connection("TEST", param) # send command to host when the connection starts.
pprint(s2f.get_connection_names())

pprint("ADD TEST_2")
s2f.add_connection("TEST_2", param, on_start="start_some_cmd")
pprint(s2f.get_connection_names())

pprint("RENOVE TEST")
s2f.remove_connection("TEST")
pprint(s2f.get_connection_names())


pprint("Show all params")
pprint(s2f.connections)
```

```python
!mkdir output_streams

from threading import Thread

output_dir = "./output_streams"

# Start all connections
for con in s2f.get_connection_names():

    # create file name or you can create mine
    name_file = s2f.create_output_file_name(con, output_dir)

    # Start process as daemon
    th = Thread(target=s2f.start, args=(con, name_file))
    th.start()

# Work 60 seconds
count = 0
while True:
    # get all connections
    for con in s2f.get_connection_names():
        status = s2f.get_status(con)
        print(status)
        count += 1
        time.sleep(1)
    if count > 60:
        break

# stop all streams
for con in s2f.get_connection_names():
    s2f.stop(con)
```
